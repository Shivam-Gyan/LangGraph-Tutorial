### @tool decorator to create a custom tool

```python
from langchain_core.tools import tool


@tool # decorator to mark this function as a tool
def addition_tool(a: int, b: int) -> int: # this not a normal function of python but a tool used by LLM to perform actions
    """A simple calculator tool that adds two numbers."""  # defination
    return a + b
   
```

```python
result = addition_tool.invoke({"a": 5, "b": 3})  # calling the tool with arguments

print(f"Addition Result: {result}")  # Output: Addition Result: 8
```

```python
print(addition_tool.description)  # Output: A simple calculator tool that adds two numbers.
print(addition_tool.name)  # Output: addition_tool
print(addition_tool.args_schema.model_json_schema())  # Output: { 'a': int, 'b': int }
print(addition_tool.args)
```
