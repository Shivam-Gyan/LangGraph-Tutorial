## Type of Tools Creation

### @tool decorator to create a custom tool

```python
from langchain_core.tools import tool


@tool # decorator to mark this function as a tool
def addition_tool(a: int, b: int) -> int: # this not a normal function of python but a tool used by LLM to perform actions
    """A simple calculator tool that adds two numbers."""  # defination
    return a + b
   
```

```python
result = addition_tool.invoke({"a": 5, "b": 3})  # calling the tool with arguments

print(f"Addition Result: {result}")  # Output: Addition Result: 8
```

```python
print(addition_tool.description)  # Output: A simple calculator tool that adds two numbers.
print(addition_tool.name)  # Output: addition_tool
print(addition_tool.args_schema.model_json_schema())  # Output: { 'a': int, 'b': int }
print(addition_tool.args)
```

### StructuredTool to define a structured tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel,Field

class MultiplySchema(BaseModel):
    x: int = Field(..., description="The first number to multiply")
    y: int = Field(..., description="The second number to multiply")


def multiply(x: int, y: int) -> int:
    """A simple calculator tool that multiplies two numbers."""
    return x * y

multiply_tool = StructuredTool.from_function(
    func=multiply,
    args_schema=MultiplySchema,
    name="multiply_tool",
    description="A tool that multiplies two numbers.",
)


result = multiply_tool.invoke({"x": 4, "y": 6})
print(f"Multiplication Result: {result}")  # Output: Multiplication Result:
```

```python
multiply_tool.args_schema.model_json_schema()  # Output: { 'x': int, 'y': int }
```

### BaseTool 

```python
from langchain.tools import BaseTool

class SubtractTool(BaseTool):
    name : str = "subtract_tool"
    description  : str= "A tool that subtracts one number from another."

    def _run(self, a: int, b: int) -> int:
        """Subtract b from a."""
        return a - b

    
subtract_tool = SubtractTool()
result = subtract_tool.invoke({"a": 10, "b": 4})
print(f"Subtraction Result: {result}")  # Output: Subtraction Result: 6
```

```python
print(subtract_tool.args)  # Output: {'a': int, 'b': int}
```
