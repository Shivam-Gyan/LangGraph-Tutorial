## How a LLM Work with tools

```python
from langchain_ollama import ChatOllama
model  = ChatOllama(model="ministral-3:3b", temperature=0)
```

```python
from langchain_core.tools import tool
```

```python
@tool 
def Add_tool(a:int, b:int)->int:

    '''This function add two numbers a and b'''
    return a+b

print(Add_tool.args_schema.model_json_schema())
```

```python
# now we have to bind this ADD_tool with Model (LLm)

LLM_with_model = model.bind_tools([Add_tool],tool_choice={"type": "function", "function": {"name": "Add_tool"}})

```

```python
from langchain_core.messages import HumanMessage

messages = [HumanMessage(content="can you use tool to add 6 and 7")]

result = LLM_with_model.invoke(messages)
# print(result)
messages.append(result)
```

```python
tool_args = result.tool_calls[0]
```

```python
tool_message = Add_tool.invoke(tool_args)
```
