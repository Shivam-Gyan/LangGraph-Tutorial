## How a LLM Work with tools

```python
from langchain_ollama import ChatOllama
model  = ChatOllama(model="ministral-3:3b", temperature=0)
```

```python
from langchain_core.tools import tool
```

```python
@tool 
def Add_tool(a:int, b:int)->int:

    '''This function add two numbers a and b'''
    return a+b

print(Add_tool.args_schema.model_json_schema())
```

```python
# now we have to bind this ADD_tool with Model (LLm)

LLM_with_model = model.bind_tools([Add_tool])

```

```python
result = LLM_with_model.invoke("hey how are you ? what is 5 + 6")
# print(result)
result
```
