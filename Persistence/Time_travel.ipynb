## Time Travel in Persistence

In LangGraph, Time Travel is essentially a coordinate system for your application's state.

* **`thread_id`**: Selects the specific **timeline** (the conversation or process).
* **`checkpoint_id`**: Selects the specific **moment** in that timeline to return to.

There is two main concept in Time Travel

#### 1. The Coordinate System

You can think of it exactly like a saved game file, as we discussed:

* **Standard Run:** You provide just the `thread_id`. LangGraph automatically finds the *latest* checkpoint.
* **Time Travel Run:** You provide `thread_id` **AND** `checkpoint_id`. LangGraph ignores the latest data and loads the state exactly as it was at that specific ID.

#### 2. The Nuance: "Forking" (Not Overwriting)

When you "start execution from that point onwards," you generally **do not delete** the history that happened after that point. Instead, you create a **new branch**.

If you go back to Step 2 and change an action:

1. The original "Step 3" and "Step 4" still exist in the database (Historical trace).
2. Your new execution creates a "Step 3 (Version B)" and continues from there.

This is powerful because you never lose data, even when debugging.
