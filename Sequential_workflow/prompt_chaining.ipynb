```python
from langgraph.graph import StateGraph,START,END
# from langchain_openai import ChatOpenAI
from langchain_google_genai import ChatGoogleGenerativeAI
from pydantic import BaseModel

from dotenv import load_dotenv
import os

load_dotenv()

# model = ChatOpenAI(api_key=os.getenv("OPENAI_KEY"), model_name="gpt-3.5-turbo", temperature=0)
model = ChatGoogleGenerativeAI(api_key=os.getenv("GOOGLE_API_KEY"), model="gemini-2.5-flash", temperature=0)
```

```python
class BLOGState(BaseModel):
    topic: str
    outline: str = ""
    draft: str = ""
    final: str = ""
```

```python
# functions of each nodes to process the state

def generate_outline(state: BLOGState) -> BLOGState:
    prompt = f"""
    Generate a detailed outline for a blog post about the following topic:

    Topic: {state.topic}
    """
    state.outline = model.invoke(prompt).content.strip()
    return state


def write_draft(state: BLOGState) -> BLOGState:
    prompt = f"""
    Using the following outline, write a detailed draft for a blog post.

    Topic: {state.topic}

    Outline:
    {state.outline}
    """
    state.draft = model.invoke(prompt).content.strip()
    return state


def finalize_blog(state: BLOGState) -> BLOGState:
    prompt = f"""
    Revise and finalize the following blog post:

    Draft:
    {state.draft}
    """
    state.final = model.invoke(prompt).content.strip()
    return state
```

```python
graph = StateGraph(BLOGState)


# add nodes
graph.add_node('generate_outline',generate_outline)
graph.add_node('write_draft',write_draft)
graph.add_node('finalize_blog',finalize_blog)


# add edges
graph.add_edge(START,'generate_outline')
graph.add_edge('generate_outline','write_draft')
graph.add_edge('write_draft','finalize_blog')
graph.add_edge('finalize_blog',END)

# compile the graph 

workflow = graph.compile()
```

```python
initial_state = {"topic": "The Impact of Artificial Intelligence on Modern Healthcare"}

final_state = workflow.invoke(initial_state)

print("Final Blog Post:\n")
print(final_state['final'])
```
